* Use the sqlitesearch library (use for larger data, or to save data in a persistent storage)
    * ex) Elasticsearch (data preserved since it's saved to disk)
* minsearch is an in-memory database. It's a bunch of Python dictionaries. (If I stop the notebook, the data will disappear)

```python
from sqlitesearch import TextSearchIndex

sqlite_index = TextSearchIndex(
    text_fields=["question", "section", "answer"],
    keyword_fields=["course"],
    db_path="faq.db"
)
```

```python
from dotenv import load_dotenv
load_dotenv()

#loads the OpenAI client
from openai import OpenAI
openai_client = OpenAI()
```

```python
from rag_helper import RAGBase

assistant = RAGBase(sqlite_index, openai_client) #index first, then the LLM client second
```
